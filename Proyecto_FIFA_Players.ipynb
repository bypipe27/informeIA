# Parte 2: Preparación de Datos

> **Conexión con Notebook 01** — Limpiamos el dataset, manejamos valores faltantes y
> creamos la variable `position_cat` (target de clasificación).

Las dos celdas siguientes ya están implementadas. **Léelas con atención.**


````python
# === [CÓDIGO COMPLETO] LIMPIEZA DE DATOS ===

print('Valores faltantes antes de limpiar:')
print(df.isnull().sum()[df.isnull().sum() > 0])

# 1. Solo jugadores con valor de mercado conocido (> 0)
df = df[df['value_eur'] > 0].copy()

# 2. Solo jugadores con posición conocida
df = df[df['player_positions'].notna()].copy()

# 3. Imputar las 6 stats principales con la mediana para no perder porteros
STATS_PRINCIPALES = ['pace', 'shooting', 'passing', 'dribbling', 'defending', 'physic']
for col in STATS_PRINCIPALES:
    df[col] = df[col].fillna(df[col].median())

# 4. Rellenar NaN restantes con la mediana
cols_num = df.select_dtypes(include=[np.number]).columns
df[cols_num] = df[cols_num].fillna(df[cols_num].median())

print(f'\nDataset limpio: {df.shape[0]:,} jugadores')
print(f'Valores faltantes restantes: {df.isnull().sum().sum()}')
````

````python
# === [CÓDIGO COMPLETO] FEATURE ENGINEERING: CATEGORÍA DE POSICIÓN ===

MAPA_POSICIONES = {
    'GK': 'GK',
    'CB': 'DEF', 'LB': 'DEF', 'RB': 'DEF', 'LWB': 'DEF', 'RWB': 'DEF',
    'CM': 'MID', 'CAM': 'MID', 'CDM': 'MID', 'LM': 'MID', 'RM': 'MID',
    'ST': 'FWD', 'CF': 'FWD', 'LW': 'FWD', 'RW': 'FWD',
    'LS': 'FWD', 'RS': 'FWD', 'LF': 'FWD', 'RF': 'FWD',
}

def categorizar_posicion(pos_str):
    if pd.isna(pos_str):
        return None
    pos_principal = pos_str.split(',')[0].strip()
    return MAPA_POSICIONES.get(pos_principal, 'MID')

df['position_cat'] = df['player_positions'].apply(categorizar_posicion)
df = df[df['position_cat'].notna()].copy()

dist = df['position_cat'].value_counts()
colores_pos = {'GK': 'gold', 'DEF': 'royalblue', 'MID': 'forestgreen', 'FWD': 'crimson'}

fig, ax = plt.subplots(figsize=(8, 5))
bars = ax.bar(dist.index, dist.values,
              color=[colores_pos[p] for p in dist.index],
              edgecolor='black', linewidth=0.8)
for bar, val in zip(bars, dist.values):
    ax.text(bar.get_x() + bar.get_width()/2., bar.get_height() + 20,
            f'{val:,}\n({val/len(df)*100:.1f}%)', ha='center', va='bottom', fontsize=11)
ax.set_xlabel('Posición', fontsize=12)
ax.set_ylabel('Número de jugadores', fontsize=12)
ax.set_title('Distribución de Jugadores por Posición', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()

print(f'\nDataset final: {len(df):,} jugadores con {len(dist)} categorías de posición')
````

# Parte 3: Regresión — ¿Cuánto vale un jugador?

> **Conexión con Notebook 02** — Aplicamos modelos de regresión para predecir el valor de
> mercado. El target es `log1p(value_eur)` para manejar la distribución sesgada.
> Para interpretar las predicciones en euros reales: `np.expm1(y_pred)`.

La celda siguiente contiene las **funciones auxiliares** y el modelo de **Regresión Lineal**
como ejemplo completo. Estúdialos antes de implementar los ejercicios.


````python
# === [CÓDIGO COMPLETO] FUNCIONES AUXILIARES + REGRESIÓN LINEAL ===

resultados_reg = []   # Acumula resultados para comparación final

def evaluar_regresion(y_real, y_pred, nombre):
    rmse = np.sqrt(mean_squared_error(y_real, y_pred))
    mae  = mean_absolute_error(y_real, y_pred)
    r2   = r2_score(y_real, y_pred)
    print(f"\n{'='*50}\n {nombre}\n{'='*50}")
    print(f'  RMSE: {rmse:.4f}')
    print(f'  MAE:  {mae:.4f}')
    print(f'  R2:   {r2:.4f}  ({r2*100:.1f}% varianza explicada)')
    resultados_reg.append({'Modelo': nombre, 'RMSE': rmse, 'MAE': mae, 'R2': r2})
    return r2


# ── Regresión Lineal (modelo base) ───────────────────────────────────────────
pipe_lr_r = Pipeline([
    ('scaler', StandardScaler()),
    ('modelo', LinearRegression())
])
pipe_lr_r.fit(X_train_r, y_train_r)
y_pred_lr_r = pipe_lr_r.predict(X_test_r)

r2_lr = evaluar_regresion(y_test_r, y_pred_lr_r, 'Regresión Lineal')

# Gráficas: Pred vs Real + Coeficientes
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

lim = [y_test_r.min() - 0.3, y_test_r.max() + 0.3]
axes[0].scatter(y_test_r, y_pred_lr_r, alpha=0.3, s=15, color='cornflowerblue')
axes[0].plot(lim, lim, 'r--', linewidth=2, label='Predicción perfecta')
axes[0].set_xlabel('log(1+value_eur) Real', fontsize=11)
axes[0].set_ylabel('log(1+value_eur) Predicho', fontsize=11)
axes[0].set_title(f'Regresión Lineal — R² = {r2_lr:.3f}', fontsize=12)
axes[0].legend()

coefs = pipe_lr_r.named_steps['modelo'].coef_
colores = ['forestgreen' if c > 0 else 'crimson' for c in coefs]
axes[1].barh(FEATURES_NUMERICAS, coefs, color=colores, edgecolor='gray', alpha=0.8)
axes[1].axvline(0, color='black', linewidth=0.8)
axes[1].set_xlabel('Coeficiente estandarizado', fontsize=11)
axes[1].set_title('Importancia de variables (Regresión Lineal)', fontsize=12)

plt.tight_layout()
plt.show()
print('Verde = aumenta el valor | Rojo = disminuye el valor')
````

Los árboles de decisión dividen el espacio de features en regiones rectangulares y predicen
el **promedio** del target en cada región. Capturan relaciones **no lineales**, a diferencia

````python
# ╔══════════════════════════════════════════════════════════╗
# ║  EJERCICIO 2: Árbol de Decisión para Regresión           ║
# ╚══════════════════════════════════════════════════════════╝

# ── TODO 2.1: Pipeline con DecisionTreeRegressor ─────────────────────────────
pipe_dt_r = Pipeline([
    ('scaler', StandardScaler()),
    ('modelo', DecisionTreeRegressor(max_depth=7, min_samples_leaf=10, random_state=RANDOM_STATE))
])
pipe_dt_r.fit(X_train_r, y_train_r)


# ── TODO 2.2: Predicciones + evaluación ──────────────────────────────────────
y_pred_dt_r = pipe_dt_r.predict(X_test_r)
r2_dt = evaluar_regresion(y_test_r, y_pred_dt_r, 'Árbol de Decisión (Regresión)')

# ── TODO 2.3: Gráfica Predicciones vs Real ───────────────────────────────────
fig, ax = plt.subplots(figsize=(7, 5))
lim = [y_test_r.min() - 0.3, y_test_r.max() + 0.3]
ax.scatter(y_test_r, y_pred_dt_r, alpha=0.3, s=15, color='cornflowerblue')
ax.plot(lim, lim, 'r--', linewidth=2, label='Predicción perfecta')
ax.legend()
ax.set_xlabel('log(1+value_eur) Real', fontsize=11)
ax.set_ylabel('log(1+value_eur) Predicho', fontsize=11)
ax.set_title(f'Árbol de Decisión — R² = {r2_dt:.3f}', fontsize=12)
plt.tight_layout()
plt.show()


# ── TODO 2.4 (Bonus): R² vs max_depth ────────────────────────────────────────
profundidades = range(1, 16)
r2_train_list = []
r2_test_list  = []

for depth in profundidades:
    pipe_temp = Pipeline([('scaler', StandardScaler()),
                          ('modelo', DecisionTreeRegressor(max_depth=depth, random_state=RANDOM_STATE))])
    # Tu código aquí: fit, predice train y test, calcula r2 para cada uno
    pipe_temp.fit(X_train_r, y_train_r)
    y_pred_train = pipe_temp.predict(X_train_r)
    y_pred_test = pipe_temp.predict(X_test_r)
    r2_train_list.append(r2_score(y_train_r, y_pred_train))
    r2_test_list.append(r2_score(y_test_r, y_pred_test))


# Grafica r2_train_list y r2_test_list vs profundidades
fig, ax = plt.subplots(figsize=(8, 5))
ax.plot(profundidades, r2_train_list, 'o-', label='R2 Train')
ax.plot(profundidades, r2_test_list, 's-', label='R2 Test')
ax.set_xlabel('max_depth')
ax.set_ylabel('R2')
ax.set_title('R2 vs max_depth (Árbol de Decisión)')
ax.legend()
plt.tight_layout()
plt.show()
````

````python
# ╔══════════════════════════════════════════════════════════╗
# ║  EJERCICIO 3: KNN para Regresión                         ║
# ╚══════════════════════════════════════════════════════════╝

# ── TODO 3.1: Pipeline con KNeighborsRegressor ───────────────────────────────
pipe_knn_r = Pipeline([
    ('scaler', StandardScaler()),
    ('modelo', KNeighborsRegressor(n_neighbors=5))
])
pipe_knn_r.fit(X_train_r, y_train_r)


# ── TODO 3.2: Predicciones + evaluación ──────────────────────────────────────
y_pred_knn_r = pipe_knn_r.predict(X_test_r)

r2_knn_r = evaluar_regresion(y_test_r, y_pred_knn_r, 'KNN Regresión (K=5)')


# ── TODO 3.3: Gráfica R² vs K ────────────────────────────────────────────────
k_valores = range(1, 31)
r2_por_k = []

for k in k_valores:
    pipe_temp = Pipeline([
        ('scaler', StandardScaler()),
        ('modelo', KNeighborsRegressor(n_neighbors=k))
    ])
    pipe_temp.fit(X_train_r, y_train_r)
    r2_por_k.append(r2_score(y_test_r, pipe_temp.predict(X_test_r)))

fig, ax = plt.subplots(figsize=(10, 5))
ax.plot(list(k_valores), r2_por_k, 'o-', color='darkorange', linewidth=2)
ax.axvline(list(k_valores)[int(np.argmax(r2_por_k))], color='red', linestyle='--', linewidth=1.5,
           label=f'Mejor K = {list(k_valores)[int(np.argmax(r2_por_k))]}')
ax.set_xlabel('Número de vecinos K', fontsize=12)
ax.set_ylabel('R² en test', fontsize=12)
ax.set_title('KNN Regresión: R² según K', fontsize=13)
ax.legend()
plt.tight_layout()
plt.show()

mejor_k_r = list(k_valores)[int(np.argmax(r2_por_k))]
print(f'Mejor K para regresión: {mejor_k_r}')
print(f'Mejor R² en test: {max(r2_por_k):.4f}')


#── BONUS: Mejor K con Validación Cruzada ────────────────────────────────────
r2_cv_por_k = []

for k in k_valores:
    pipe_cv = Pipeline([
        ('scaler', StandardScaler()),
        ('modelo', KNeighborsRegressor(n_neighbors=k))
    ])
    cv_scores = cross_val_score(pipe_cv, X_reg, y_reg, cv=5, scoring='r2', n_jobs=-1)
    r2_cv_por_k.append(cv_scores.mean())

fig, ax = plt.subplots(figsize=(10, 5))
ax.plot(list(k_valores), r2_cv_por_k, 'o-', color='steelblue', linewidth=2)
ax.axvline(list(k_valores)[int(np.argmax(r2_cv_por_k))], color='red', linestyle='--', linewidth=1.5,
           label=f'Mejor K CV = {list(k_valores)[int(np.argmax(r2_cv_por_k))]}')
ax.set_xlabel('Número de vecinos K', fontsize=12)
ax.set_ylabel('R² medio en validación cruzada', fontsize=12)
ax.set_title('KNN Regresión: R² medio por K usando CV', fontsize=13)
ax.legend()
plt.tight_layout()
plt.show()

mejor_k_cv = list(k_valores)[int(np.argmax(r2_cv_por_k))]
print(f'Mejor K por validación cruzada: {mejor_k_cv}')
print(f'Mejor R² CV medio: {max(r2_cv_por_k):.4f}')
````

# Bonus — Conversión de Predicciones a Euros Reales

Durante el entrenamiento de los modelos de regresión, la variable objetivo (`value_eur`) fue transformada mediante `np.log1p()` con el fin de reducir la asimetría de los datos y mejorar el desempeño de los algoritmos. Sin embargo, las predicciones obtenidas se encuentran en escala logarítmica, lo que dificulta su interpretación directa.

Para recuperar los valores monetarios originales, se aplica la transformación inversa `np.expm1()`, que permite convertir tanto las predicciones como los valores reales nuevamente a euros. De esta forma es posible evaluar el error del modelo en una unidad más comprensible y comparar directamente el valor estimado de los jugadores con su valor real de mercado.


````python
# ╔══════════════════════════════════════════════════════════╗
# ║ BONUS: Comparación de Modelos en Euros Reales           ║
# ╚══════════════════════════════════════════════════════════╝

# Recuperar escala original (€)
y_real_eur = np.expm1(y_test_r)

y_pred_lr_eur  = np.expm1(y_pred_lr_r)
y_pred_dt_eur  = np.expm1(y_pred_dt_r)
y_pred_knn_eur = np.expm1(y_pred_knn_r)

# Métricas en euros
resultados_eur = pd.DataFrame([
    {
        'Modelo': 'Regresión Lineal',
        'RMSE (€)': np.sqrt(mean_squared_error(y_real_eur, y_pred_lr_eur)),
        'MAE (€)': mean_absolute_error(y_real_eur, y_pred_lr_eur)
    },
    {
        'Modelo': 'Árbol de Decisión',
        'RMSE (€)': np.sqrt(mean_squared_error(y_real_eur, y_pred_dt_eur)),
        'MAE (€)': mean_absolute_error(y_real_eur, y_pred_dt_eur)
    },
    {
        'Modelo': 'KNN (K=5)',
        'RMSE (€)': np.sqrt(mean_squared_error(y_real_eur, y_pred_knn_eur)),
        'MAE (€)': mean_absolute_error(y_real_eur, y_pred_knn_eur)
    }
])

resultados_eur = resultados_eur.sort_values('RMSE (€)')

print('=' * 70)
print('COMPARACIÓN DE MODELOS EN EUROS REALES')
print('=' * 70)
print(resultados_eur.round(2))

comparacion = pd.DataFrame({
    'Valor Real (€)': y_real_eur.values,
    'Regresión Lineal (€)': y_pred_lr_eur,
    'Árbol de Decisión (€)': y_pred_dt_eur,
    'KNN (€)': y_pred_knn_eur
})

print('\nEjemplos de predicciones:')
print(comparacion.sample(10, random_state=42).round(0))
resultados_eur['RMSE (€)'] = resultados_eur['RMSE (€)'].map(lambda x: f'€{x:,.0f}')
resultados_eur['MAE (€)'] = resultados_eur['MAE (€)'].map(lambda x: f'€{x:,.0f}')

print(resultados_eur)
````

## Bonus — Análisis de Jugadores Mal Clasificados

Para identificar los errores más frecuentes se analizaron las matrices de confusión de los modelos de clasificación. Los resultados muestran que los principales errores se concentran entre las posiciones **DEF**, **MID** y **FWD**, mientras que los **GK** fueron clasificados correctamente en prácticamente todos los casos.

### Principales confusiones observadas

**1. Defensas clasificados como Mediocampistas (DEF → MID)**

Este fue uno de los errores más frecuentes en los tres modelos. Por ejemplo, el Random Forest confundió 723 defensas como mediocampistas, mientras que el Árbol de Decisión confundió 1158.

Esta situación tiene sentido porque algunos defensas modernos poseen buenas estadísticas de pase, control de balón y visión de juego, características que también son comunes en los mediocampistas.

**2. Mediocampistas clasificados como Defensas (MID → DEF)**

También se observó una cantidad importante de errores en sentido contrario. Muchos mediocampistas defensivos (CDM) presentan atributos de defensa y físico muy similares a los de un defensa central o lateral, lo que dificulta la separación entre ambas posiciones.

**3. Delanteros clasificados como Mediocampistas (FWD → MID)**

La mayor confusión relacionada con los delanteros ocurre con los mediocampistas ofensivos. En el caso de Random Forest se confundieron 1050 delanteros como mediocampistas.

Esto sucede porque jugadores ofensivos modernos comparten estadísticas elevadas en disparo, regate y pase, haciendo que los límites entre ambas posiciones sean menos claros.

### Posición más fácil de identificar

Los porteros (**GK**) fueron identificados correctamente por todos los modelos. En las matrices de confusión se observa una clasificación perfecta, sin errores significativos.

Esto ocurre porque sus atributos son muy diferentes a los de los jugadores de campo. Variables como ritmo, disparo, regate y pase presentan valores característicos que permiten distinguirlos fácilmente.

### Conclusión

Los errores de clasificación observados reflejan situaciones que también ocurren en el fútbol real. Las posiciones DEF, MID y FWD forman un continuo de características donde existen perfiles híbridos que comparten atributos similares. Por el contrario, los porteros presentan un perfil estadístico único que facilita su identificación. El modelo que mejor manejó estas ambigüedades fue Random Forest, alcanzando una precisión cercana al 90% y reduciendo la cantidad de confusiones entre posiciones vecinas.


# Bonus — Gradient Boosting vs Random Forest

Además de los modelos evaluados anteriormente, se implementó un clasificador basado en **Gradient Boosting** para comparar su desempeño con el modelo de **Random Forest**.

Mientras que Random Forest construye múltiples árboles de manera independiente y combina sus predicciones mediante votación, Gradient Boosting construye árboles secuencialmente, donde cada nuevo árbol intenta corregir los errores cometidos por los anteriores.

Esta estrategia suele producir modelos más precisos, aunque con un mayor costo computacional y una mayor sensibilidad a los hiperparámetros. El objetivo es determinar si Gradient Boosting puede mejorar la capacidad de clasificación de las posiciones de los jugadores respecto al Random Forest.

````python
# ╔══════════════════════════════════════════════════════════╗
# ║ BONUS: Gradient Boosting Classifier                     ║
# ╚══════════════════════════════════════════════════════════╝

from sklearn.ensemble import GradientBoostingClassifier

# ── Entrenamiento ─────────────────────────────────────────
pipe_gb_c = Pipeline([
    ('scaler', StandardScaler()),
    ('modelo', GradientBoostingClassifier(
        n_estimators=100,
        learning_rate=0.1,
        max_depth=3,
        random_state=RANDOM_STATE
    ))
])

pipe_gb_c.fit(X_train_c, y_train_c)

# ── Predicciones + evaluación ─────────────────────────────
y_pred_gb_c = pipe_gb_c.predict(X_test_c)

acc_gb_c = evaluar_clasificacion(
    y_test_c,
    y_pred_gb_c,
    'Gradient Boosting'
)

# ── Matriz de confusión ───────────────────────────────────
fig, ax = plt.subplots(figsize=(7, 6))

cm_gb = confusion_matrix(
    y_test_c,
    y_pred_gb_c,
    labels=clases
)

sns.heatmap(
    cm_gb,
    annot=True,
    fmt='d',
    cmap='Purples',
    ax=ax,
    xticklabels=clases,
    yticklabels=clases,
    annot_kws={'size': 13}
)

ax.set_xlabel('Predicción', fontsize=12)
ax.set_ylabel('Real', fontsize=12)
ax.set_title(
    f'Matriz de Confusión — Gradient Boosting\nAccuracy: {acc_gb_c:.3f}',
    fontsize=13
)

plt.tight_layout()
plt.show()

# ── Comparación con Random Forest ─────────────────────────
comparacion_ensambles = pd.DataFrame({
    'Modelo': [
        'Random Forest',
        'Gradient Boosting'
    ],
    'Accuracy': [
        acc_rf_c,
        acc_gb_c
    ]
})

comparacion_ensambles = comparacion_ensambles.sort_values(
    'Accuracy',
    ascending=False
)

print('\n' + '='*60)
print('COMPARACIÓN: RANDOM FOREST vs GRADIENT BOOSTING')
print('='*60)
print(comparacion_ensambles)

# ── Mejor modelo ──────────────────────────────────────────
mejor_modelo = comparacion_ensambles.iloc[0]

print('\nMejor modelo:')
print(f"{mejor_modelo['Modelo']} (Accuracy = {mejor_modelo['Accuracy']:.4f})")
````
